````python
import csv
````

Importing the `csv` module gives is a great amount of power to 
open the csv file and let the module do the rest.  Specifically, 
though, it is an important thing for us to give the module a hint 
about the CSV  file structure, possibly hinting at
whethere the first line is a header and of the delimiter is a 
comma (`,`).  **NOTE:** While it may seem obvious that the "C"
in "CSV" implies that the delimiter is a comma, it is important to 
verify this -- sometime it is a space, tab or something else 
altogether (`|` is often used, for example.) 


Looking at the first few lines of the file reveals some truths:

```csv
time,air
2013-01-01 00:00:00,260.1
2013-01-01 06:00:00,259.4
2013-01-01 12:00:00,260.0
...

```

1. the file has a header line indicating the column names `time` and `air`,
2. the `,` comma indeed delimits the fields.

Lucky for us, the header can be extracted automatically with the [`csv.DictReader()`](https://docs.python.org/3/library/csv.html#csv.DictReader) simplifying 
our lives greatly.

````python
with open("data/Yellowstone-National-Park_249.411484_44.423691_.csv") as f:
    reader = csv.DictReader(f)
    for line in reader:
        print(line)
        break  # only consume the first line
````

We can see we have a dictionary with keys `time` and `air` along with
the corresponding values for each.

To store the data, we will loop through every line of the file
and put the corresponding `DictReader` into a list.  This will
produce a list of dictionaries:

````python
data = []
with open("data/Yellowstone-National-Park_249.411484_44.423691_.csv") as f:
    reader = csv.DictReader(f)
    for line in reader:
        data.append(line)
````

Notice the fist 10 items of our `data` list:

````python
data[:10]
````

And if we want to access the 10th line of the data:

````python
data[10]['time'], data[10]['air']
````
